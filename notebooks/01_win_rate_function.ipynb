# Milestone 1 – Exercise 1: passing data into a function

Goal: rewrite `calculate_win_rate` so it receives the matches as an argument
instead of reading the global variable `matches`.

```python
import pandas as pd                         

url = "https://raw.githubusercontent.com/farhadGithub/tennis-atp-data/main/data/raw/atp_matches_2024.csv"

matches = pd.read_csv(url)
print(f"Loaded {len(matches)} matches")
```

## Part A

Write `calculate_win_rate(data, player, surface)`. Inside the function, use `data`, never `matches`.

```python
# Write your function here
def calculate_win_rate(data, player, surface):
    # Filter player wins and losses
    player_wins = data[data["winner_name"] == player]
    player_losses = data[data["loser_name"] == player]

    # Filter by surface
    surface_wins = player_wins[player_wins["surface"] == surface]
    surface_losses = player_losses[player_losses["surface"] == surface]

    # Count matches
    total_matches = len(surface_wins) + len(surface_losses)

    # Handle missing data
    if total_matches == 0:
        return None

    # Calculate win rate
    win_rate = (len(surface_wins) / total_matches) * 100
    return win_rate
```

```python
# Check: this should print 81.0
print(f"{calculate_win_rate(matches, 'Carlos Alcaraz', 'Clay'):.1f}")
```

2. Call your function with `first_half` to get Alcaraz's clay win rate in the first half of 2024.
3. Compare it with the full-year result. Is it different? Why?

```python
# Write Part B here
first_half = matches[matches["tourney_date"] < 20240701]
print(f"Matches in first half: {len(first_half)}")

first_half_rate = calculate_win_rate(first_half, "Carlos Alcaraz", "Clay")
full_year_rate = calculate_win_rate(matches, "Carlos Alcaraz", "Clay")

print(f"Alcaraz on clay, first half: {first_half_rate:.1f}%")
print(f"Alcaraz on clay, full year:  {full_year_rate:.1f}%")
```

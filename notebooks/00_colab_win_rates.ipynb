```python

import pandas as pd

url = "https://raw.githubusercontent.com/farhadGithub/tennis-atp-data/main/data/raw/atp_matches_2024.csv"

matches = pd.read_csv(url)
#Alcaraz
alcaraz_wins = matches[matches["winner_name"]== "Carlos Alcaraz"]
print(f"Carlos Alcaraz has won:{len(alcaraz_wins)} matches")

alcaraz_losses = matches[matches["loser_name"] == "Carlos Alcaraz"]

alcaraz_matches = len(alcaraz_wins) + len(alcaraz_losses)

alcaraz_win_rate = (len(alcaraz_wins)/alcaraz_matches)*100
print(f"Carlos Alcaraz has:{alcaraz_win_rate:.1f} win rate")

clay_wins = alcaraz_wins[alcaraz_wins["surface"] == "Clay"]
print(len(clay_wins))

clay_losses = alcaraz_losses[alcaraz_losses["surface"] == "Clay"]
print(len(clay_losses))

alcaraz_clay_matches_total= len(clay_wins) + len(clay_losses)

alcaraz_clay_win_rate=(len(clay_wins)/alcaraz_clay_matches_total) * 100
print(f"Carlos Alcaraz has a Win Rate of:{alcaraz_clay_win_rate:.1f}% on Clay surface")

print(f"Wins: {len(alcaraz_wins)}")
print(f"Losses: {len(alcaraz_losses)}")
print(f"Win rate: {alcaraz_win_rate:.1f}%")

print(matches[["winner_name","loser_name","surface","score"]].head(10))

print(f"Total matches:{len(matches)}")
print(matches.columns)
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/farhadGithub/tennis-atp-data/main/data/raw/atp_matches_2024.csv"

matches = pd.read_csv(url)

def calculate_win_rate(player,surface):
  #player
  player_losses = matches[matches["loser_name"] == player]
  player_wins = matches[matches["winner_name"] == player]
  #surface
  surface_wins = player_wins[player_wins["surface"] == surface]
  surface_losses = player_losses[player_losses["surface"] == surface]
  #win rate
  total_matches = len(surface_wins) + len(surface_losses)
  if total_matches == 0:
    return None
  win_rate= (len(surface_wins)/total_matches) * 100
  return win_rate

players = ["Carlos Alcaraz", "Jannik Sinner", "Novak Djokovic", "Roger Federer"]
for player in players:
  result = calculate_win_rate(player,"Clay")
  if result is None:
   print(f"{player}: No data available")
  else:
     print(f"{player}:{result:.1f}%")

```
